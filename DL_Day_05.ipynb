```python
# Problem with Perceptron Trick/Rule
'''
We can't say that the line we get using perceptron trick is the ultimate line for the data to classify every point.
So the value we get of A,B and C or W1,W2 and B is not the final value to classify every point.
We can't quantify the line that we get using perceptron trick means we can't say how good our line is performence wise.
Also we can face convergence issue like the model selecting the same point to measure if it's correctly classified or not and as a result the wrong line might be produced and it might not be changed.
So, in perceptron trick we got two issue-
  - Quantify our line
  - Convergence issue
To resolve these issue we use Loss Function.
'''
```

```python
# Loss Function
'''
Loss function is function which tells us how good or bad our model is performing.
Loss function give us a number which tells us the error of our model. It tells us which line has the less error.
The line with the less error will be our final line and the value of W1,W2 and b of that line will be our final value.
So, loss function tells us the which is the best line for our problem and model.
'''
```

```python
'''
There are multiple loss function according to every problem.
But in perceptron we use a loss function where we use the line equation to calculate the loss function instead of calculating the distance of the problem. Becuase it is proportional to distance. The point has more distance also has the greater value of the equation if we put the coordinates of the piont into that line equation.
So the loss function for Perceptron is-
L(w1,w2,b) = 1/n Sum(L(y_i, f(x_i))) + alpha*R(w1,w2)
n = number of rows in data
y_i = is the output column data in our dataset

alpha*R(w1,s2)----> is for regularization

L(y_i,f(x_i)) = max(0,-y_i*f(x_i))
f(x_i) = w1*x1 + w2*x2 + b ----> actually 'z'

So, our objective is to find the value of w1,w2 and b for which we will get the minimum value of the loss function. For this we will use gradiet descent.
'''
```

```python
'''
For finding the minimum value of w1,w2 and b for which we can get the minimum value of L(w1,w2,b) we will use gradiet descent.
So for that-
we will randomly initialize w1,w2 and b.
  for i in epochs:
    w1 = w1 + learning-rate*dl/dw1
    w2 = w2 + learning-rate*dl/dw2
    b = b + learning-rate*dl/db
  return w1,w2,b

for differentiate dl/dw1:
  we will use chain rule-
  dl/dw1 = dl/df(x_i)*df(x_i)/dw1

  dl/df(x_i) = {0 if y_i*f(x_i)>=0 and -y_i if y_i*f(x_i)<0}
  df(x_i1) = w1*x1 + w2*x2 + b
  df(x_i1)/dw1 = x1

  dl/dw1 = {0 if y_i*f(x_i)>=0 and -y_i*x_i1 if y_i*f(x_i)<0}
  dl/dw2 = {0 if y_i*f(x_i)>=0 and -y_i*x_i2 if y_i*f(x_i)<0}
  dl/db = {0 if y_i*f(x_i)>=0 and -y_i if y_i*f(x_i)<0}


'''
```

```python
from sklearn.datasets import make_classification
import numpy as np
X,y = make_classification(n_samples = 100, n_features=2,n_informative=1, n_redundant=0,
                          n_classes = 2, n_clusters_per_class=1, random_state=41, hypercube=False, class_sep=15)
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10,6))
plt.scatter(X[:,0], X[:,1], c=y, cmap='winter', s=100)
```

```python
def perceptron(X,y):
  w1=w2=b=1
  lr = 0.01

  for j in range(10000):
    for i in range(X.shape[0]):
      # check condition
      z = w1*X[i][0] + w2*X[i][1] + b

      if z*y[i] <0:
        w1 = w1 + lr*y[i]*X[i][0]
        w2 = w2 + lr*y[i]*X[i][1]
        b = b + lr*y[i]
  return w1,w2,b
```

```python
w1,w2,b = perceptron(X,y)
```

```python
m = -(w1/w2)
c = -(b/w2)
print(m,c)
```

```python
x_input = np.linspace(-3,3,100)
y_input = m*x_input + c
```

```python
plt.figure(figsize=(10,6))
plt.plot(x_input,y_input, color='red', linewidth=3)
plt.scatter(X[:,0], X[:,1], c=y, cmap='winter', s=100)
plt.ylim(-3,2)
```

```python
'''
Perceptrons have fundamental limitations in deep learning, specifically their inability to handle non-linearly separable data.

Linear vs. Non-linear:
  While Perceptrons excel at classifying data that can be separated by a single straight line (like AND and OR logic gates), they fail completely when faced with data requiring a more complex, non-linear boundary, such as the XOR problem.
So, perceptron is a linear model which can't work on non-linear data and never converged.
'''
```
