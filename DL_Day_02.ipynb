```python
'''
Types of Neural Network-
  There are many types of neural network. But there are 5 neural networks which are mostly used.
  1. Multi-layer perceptron(MLP)
    - A multilayer perceptron is a basic type of artificial neural network that passes information forward through fully connected layers to learn complex patterns.
    - Used mostly in supervised learning
    - Can be used for both classification and regression problems.
    - Useful to capture non-linear relationship
    - Contains input, output and multiple hidden perceptron layer
  2. Convolutional Neural Network(CNN)
    - A convolutional neural network (CNN) is a deep learning algorithm designed to process and analyze visual data, like images and videos, by preserving spatial relationships.
    - Contains Input layer, convolutional layer, ReLU, Pooling layer, Fully connected layer, Output layer. (A convolutional layer is the primary building block of a Convolution Neural Network that slides small filters across input data to extract visual patterns like edges, textures, and shapes.)
    - Use mostly for image classification or image related predicitons means use mostly on image data
  3. Recurrent Neural Network(RNN)
    - A recurrent neural network (RNN) is a type of artificial neural network designed to process sequential data by using internal loops to pass information from one step to the next.
    - Mostly used in LSTM and NLP
    - In this neural network we just don't pass the data linearly form starting layer to end layer we can get feedback from the last layer to the first layer and again work on it which actually does the loop.
    - Contains Hidden states, Feedback loops, Shared weights.
  4. Auto Encoders(AE)
    - An autoencoder is an unsupervised neural network that compresses input data into a smaller latent representation and then reconstructs it to match the original input as closely as possible.
    - Contains Encoder, Bottleneck, Decoder, Reconsruction loss
    - In this neural network the input layer and the output layer contains same number of node but the hidden layers contains less number of nodes to compress the data.
  5. Generative Adversarial Networks(GAN)
    - A generative adversarial network (GAN) is a type of neural network that consists of two main components: a generator and a discriminator.
    - Generative Adversarial Networks (GANs) are a class of machine learning frameworks introduced by Ian Goodfellow in 2014 where two neural networks compete in a zero-sum game to generate realistic synthetic data.
    - The Generator: Takes random noise as input and creates fake data (such as images, audio, or text) trying to mimic real-world datasets.
    - The Discriminator: Acts as a binary classifier that evaluates inputs and tries to distinguish whether they are real data from the training set or fake samples produced by the generator.
    - Common types of GAN are - Vanilla GAN(The basic architecture using simple multilayer perceptrons), Conditional GAN (Adds specific labels or conditional parameters to guide data generation), Deep Convolutional GAN (DCGAN)( Integrates convolutional neural networks for stable and high-quality image synthesis) etc.


Application areas of Deep Learning-
  1. Chatbots, Recommendaitons engine etc.
  2. Self-driving cars
  3. Game playing agents
  4. Virtual assistants
  5. Image colorization
  6. Adding audio or mute videos
  7. Image caption generation
  8. Text translation
  9. Pixel restoraion
  10. Object detection and Identification
  11. Generate image, audio, video etc.
'''
```
