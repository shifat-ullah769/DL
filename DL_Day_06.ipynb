```python
# Multi Layer Perceptron(MLP)
'''
A Multi-Layer Perceptron (MLP) is a foundational type of artificial neural network made of multiple layers of nodes that processes data to solve complex classification and regression tasks.

Core Structure-
  An MLP is a feedforward network, meaning information flows strictly in one direction—from start to finish, with no loops. It features three types of layers:
    Input layer: Receives the original raw data features and passes them forward.
    Hidden layer(s): One or more middle layers sitting between the input and output. This is where the heavy lifting happens, using weighted connections and activation functions to identify patterns.
    Output layer: Produces the final prediction or classification result.
  In a standard MLP, the network is fully connected. Every single node or neuron in one layer links to every single neuron in the next layer.

**Forward propagation**: Data enters the input layer, moves through the hidden layers where inputs are multiplied by weights, adds a bias, and passes through a non-linear activation function (like ReLU or sigmoid) to generate an output.
**Backpropagation**: The network compares its prediction to the actual target answer to measure the error. It then works backward through the network, using optimization algorithms (like gradient descent) to tweak and update the weights and biases so it makes fewer mistakes next time.

Before trainig and designing a neural network we have to know how much trainable parameters we have to use.
Trainable parameters means how many weights and biases we have to use. We should know this after watching the structure of our neural network.

Notation for denoting bias, output and bias-
1) bias ---> b_ij where i=layers and j=node
2) output ---> o_ij where i=layers and j=node
3) weights ---> w_ij^k where i=from which node it coming out and j=which node it is going in and k=which layer it is in.


** The main problem with perceptron is it can't work with non-linear decision boundaries/it can't capture nonlinearity because it is a linear model. So for solving this problem we use multi-layer perceptron. **

'''
```
